```python
import seaborn as sns
import matplotlib.pyplot as plt

# data
marks = [45, 50, 52, 55, 58, 60, 62, 65, 68, 70, 72, 75, 78, 80, 95]

# Create box plot
sns.boxplot(y=marks)

# Add title
plt.title("Box Plot of Student Marks")

# Display plot
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Data
hours = [1, 2, 3, 4, 5, 6, 7, 8]
marks = [40, 45, 50, 60, 65, 72, 80, 88]

# Create scatter plot
sns.scatterplot(x=hours, y=marks)

plt.title("Hours Studied vs Marks")
plt.xlabel("Hours Studied")
plt.ylabel("Marks")

plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Data
marks = [45, 50, 52, 55, 58, 60, 62, 65, 68, 70,
         72, 75, 78, 80, 82, 85, 88, 90, 92, 95]

# Create histogram
sns.histplot(marks, bins=5)

plt.title("Distribution of Student Marks")
plt.xlabel("Marks")
plt.ylabel("Number of Students")

plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Data
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
sales = [100, 120, 110, 150, 170, 160]

# Create line chart
sns.lineplot(x=months, y=sales, marker='o')

plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Data
activities = ["Study", "Sleep", "Exercise", "Entertainment"]
hours = [6, 8, 2, 4]

# Create pie chart
plt.pie(hours, labels=activities, autopct="%1.1f%%")

plt.title("Daily Time Distribution")

plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

students = ["A", "B", "C", "D", "E"]
marks = [70, 85, 60, 90, 75]

sns.barplot(x=students, y=marks)

plt.title("Student Marks")
plt.xlabel("Students")
plt.ylabel("Marks")

plt.show()
```
