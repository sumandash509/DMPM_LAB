```python
# Find mean of data points

data = [10, 20, 30, 40, 50]
total = 0

for i in data:
    total = total + i
mean = total / len(data)

print("Mean =", mean)
```

```python
# Program to find average using function

def average(data):
    return sum(data) / len(data)

data = [10, 20, 30, 40, 50]

print("Average =", average(data))
```

```python
# Program to find median without using functions

data = [10, 20, 30, 40, 50]

data.sort()
n = len(data)

if n % 2 == 0:
    median = (data[n//2 - 1] + data[n//2]) / 2
else:
    median = data[n//2]

print("Median =", median)
```

```python
# Program to find median using function

def median(data):
    data.sort()
    n = len(data)

    if n % 2 == 0:
        return (data[n//2 - 1] + data[n//2]) / 2
    else:
        return data[n//2]

data = [10, 20, 30, 40, 50]

print("Median =", median(data))
```

```python
# Program to find mode without using function

data = [1, 2, 2, 3, 4, 2, 5]

mode = data[0]
max_count = 0

for i in data:
    count = 0
    for j in data:
        if i == j:
            count += 1
    if count > max_count:
        max_count = count
        mode = i

print("Mode =", mode)
```

```python
# Program to find mode using function

def mode(data):
    max_count = 0
    mode_value = data[0]

    for i in data:
        count = data.count(i)
        if count > max_count:
            max_count = count
            mode_value = i

    return mode_value

data = [1, 2, 2, 3, 4, 2, 5]

print("Mode =", mode(data))
```

```python
# Program to find variance without using function

data = [10, 20, 30, 40, 50]

n = len(data)
total = 0

for i in data:
    total += i

mean = total / n

variance = 0

for i in data:
    variance += (i - mean) ** 2

variance = variance / n

print("Variance =", variance)
```

```python
# Program to find variance using function

def variance(data):
    mean = sum(data) / len(data)
    var = 0

    for i in data:
        var += (i - mean) ** 2

    return var / len(data)

data = [10, 20, 30, 40, 50]

print("Variance =", variance(data))
```

```python
# Program to find standard deviation without using function

data = [10, 20, 30, 40, 50]

n = len(data)
mean = sum(data) / n

variance = 0
for i in data:
    variance += (i - mean) ** 2

variance = variance / n
std = variance ** 0.5

print("Standard Deviation =", std)
```

```python
# Program to find standard deviation using function

def standard_deviation(data):
    mean = sum(data) / len(data)

    variance = 0
    for i in data:
        variance += (i - mean) ** 2

    variance = variance / len(data)
    return variance ** 0.5

data = [10, 20, 30, 40, 50]

print("Standard Deviation =", standard_deviation(data))
```

```python
# Program to find variance and standard deviation

n = int(input("Enter the number of data points: "))

data = []

for i in range(n):
    x = float(input("Enter data point: "))
    data.append(x)

mean = sum(data) / n

variance = 0

for i in data:
    variance += (i - mean) ** 2

variance = variance / n
std = variance ** 0.5

print("Variance =", variance)
print("Standard Deviation =", std)
```
